```python
# import libraries
import pandas as pd
import numpy as np
```

```python
# import dataset
df = pd.read_csv('/content/drive/MyDrive/survey_results_public.csv')
df1 = pd.DataFrame(df)
df1
```

```python
# filter by USA
df_usa = df1[df1['Country'] == 'United States of America']
display(df_usa)
```

```python
# remove incomplete data in ConvertedCompYearly
df_usa_cleaned = df_usa.dropna(subset=['ConvertedCompYearly'])
display(df_usa_cleaned)
```

```python
# filter by employed
df_employed_cleaned = df_usa_cleaned[(df_usa_cleaned['Employment'] == 'Employed')]
df_employed_cleaned.head()
```

```python
# remove extreme outliers
df_employed_cleaned = df_usa_cleaned[
    (df_usa_cleaned['ConvertedCompYearly'] >= 30000) &
    (df_usa_cleaned['ConvertedCompYearly'] <= 500000)
]
display(df_employed_cleaned.head())
```

```python
# filter years coding between 1 and 30 years
df_employed_cleaned = df_employed_cleaned[
    (df_employed_cleaned['YearsCode'] >= 1) &
    (df_employed_cleaned['YearsCode'] <= 30)
]
df_employed_cleaned
```

```python
import os

# define the directory path
output_dir = '/content/Drive/MyDrive/Colab Notebooks'

# create the directory if it does not exist
os.makedirs(output_dir, exist_ok=True)

# export as CSV to google drive
df_employed_cleaned.to_csv(os.path.join(output_dir, 'survey_USA_cleaned.csv'))
```

```python
# create salary quartiles
df_employed_cleaned['SalaryQuartile'] = pd.qcut(
    df_employed_cleaned['ConvertedCompYearly'],
    q=4,
    labels=['Q1 (lowest 25%)', 'Q2 (25%-50%)', 'Q3 (50%-75%)', 'Q4 (top 25%)']
)

# display the count of each quartile
print(df_employed_cleaned['SalaryQuartile'].value_counts().sort_index())
display(df_employed_cleaned.head())
```

```python
import plotly.express as px
#label x and y axis for box plot
fig = px.box(df_employed_cleaned, x='SalaryQuartile', y='ConvertedCompYearly')
fig.update_layout(
    title='Salary Distribution by Quartile in the USA (Employed Developers)',
    xaxis_title='Salary Quartile',
    yaxis_title='Converted Yearly Salary (USD)'
)
# order of the x-axis categories
fig.update_xaxes(
    categoryorder='array',
    categoryarray=['Q1 (lowest 25%)', 'Q2 (25%-50%)', 'Q3 (50%-75%)', 'Q4 (top 25%)']
)
fig.show()
```

```python
# create a descriptive statistics table
job_satisfaction_by_quartile = df_employed_cleaned.groupby('SalaryQuartile')['JobSat'].describe()
display(job_satisfaction_by_quartile)
```

```python
# calculate standard deviation of job satisfaction for each quartile
std_dev = df_employed_cleaned.groupby('SalaryQuartile')['JobSat'].std()
print(std_dev)
```

```python
# calculate range of job satisfaction for each quartile
range = df_employed_cleaned.groupby('SalaryQuartile')['JobSat'].max() - df_employed_cleaned.groupby('SalaryQuartile')['JobSat'].min()
print(range)
```

```python
# calculate mean for job satisfaction for each quartile
mean = df_employed_cleaned.groupby('SalaryQuartile')['JobSat'].mean()
print(mean)
```

```python
# test correlation between salary and job satisfaction using numpy basic correlation coefficient
df_correlation = df_employed_cleaned[['ConvertedCompYearly', 'JobSat']].copy()
df_correlation_cleaned = df_correlation.dropna()
print("Cleaned DataFrame for correlation:\n", df_correlation_cleaned.head())
```

```python
correlation_coefficient = df_correlation_cleaned['ConvertedCompYearly'].corr(df_correlation_cleaned['JobSat'])
print(f"Pearson correlation coefficient between ConvertedCompYearly and JobSat: {correlation_coefficient}")
```

```python
import plotly.express as px

fig = px.scatter(df_correlation_cleaned, x='ConvertedCompYearly', y='JobSat',
                 title='Scatter Plot of Salary vs. Job Satisfaction in the USA')
fig.update_xaxes(title_text='Converted Yearly Salary (USD)')
fig.update_yaxes(title_text='Job Satisfaction (1-15)')
fig.show()
```

```python
import plotly.express as px

# prepare the data for the stacked bar chart
satisfaction_by_quartile = df_employed_cleaned.groupby(['SalaryQuartile', 'JobSat'], observed=False).size().reset_index(name='Count')

# create the stacked bar chart
fig = px.bar(satisfaction_by_quartile,
             x='SalaryQuartile',
             y='Count',
             color='JobSat',
             title='Job Satisfaction Levels by Salary Quartile',
             labels={'SalaryQuartile': 'Salary Quartile', 'Count': 'Number of Respondents', 'JobSat': 'Job Satisfaction'})

# ensure the order of SalaryQuartile is correct on the x-axis
fig.update_xaxes(
    categoryorder='array',
    categoryarray=['Q1 (lowest 25%)', 'Q2 (25%-50%)', 'Q3 (50%-75%)', 'Q4 (top 25%)']
)

fig.show()
```

```python
# compare satisfaction ratings across quartiles using chi-square test
df_for_contingency = df_employed_cleaned.dropna(subset=['SalaryQuartile', 'JobSat'])
contingency_table = pd.crosstab(df_for_contingency['SalaryQuartile'], df_for_contingency['JobSat'])
display(contingency_table)
```

```python
from scipy.stats import chi2_contingency

chi2, p_value, dof, expected = chi2_contingency(contingency_table)

print(f"Chi-square statistic: {chi2}")
print(f"P-value: {p_value}")
print(f"Degrees of freedom: {dof}")
print("Expected frequencies:\n", expected)
```
